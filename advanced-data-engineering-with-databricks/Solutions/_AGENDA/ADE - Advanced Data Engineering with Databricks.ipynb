# Advanced Data Engineering with Databricks

The Advanced Data Engineering with Databricks (ADEWD) course is designed to prepare students for the Databricks Certified Professional Data Engineer certification exam. The content for this course consists of the Professional-level modules of the Data Engineer Learning Path, and can be delivered as an instructor-led training (ILT) or self-paced (SP) course.

## Course Agenda
The following modules are part of the Data Engineer Learning Path by Databricks Academy.
| # | Module Name |
| --- | --- |
| 1 | [Incremental Processing with Spark Structured Streaming]($../ADE 1 - Incremental Processing with Spark Structured Streaming/ADE 1.0 - Module Introduction) |
| 2 | [Streaming ETL Patterns with DLT]($../ADE 2 - Streaming ETL Patterns with DLT/ADE 2.0 - Module Introduction) |
| 3 | [Data Privacy and Governance Patterns]($../ADE 3 - Data Privacy and Governance/ADE 3.0 - Module Introduction) |
| 4 | [Performance Optimization with Spark and Delta Lake]($../ADE 4 - Performance Optimization/ADE 4.0 - Module Introduction) |
| 5 | [SWE Patterns for Delta Live Tables]($../ADE 5 - SWE Patterns for DLT/ADE 5.0 - Module Introduction) |
| 6 | [Automate Production Workflows]($../ADE 6 - Automate Production Workflows/ADE 6.0 - Module Introduction) |
