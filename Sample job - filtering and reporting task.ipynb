```python
%sql
-- Let's take a look at loan volume and count by loan status with this SQL query
select
  l.loan_status,
  zip_code,
  count(l.loan_amnt) as count,
  sum(l.loan_amnt) as volume
from loans l
group by loan_status, zip_code
order by 4 desc
```
