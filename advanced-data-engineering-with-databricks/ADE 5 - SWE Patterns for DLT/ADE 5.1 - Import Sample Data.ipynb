# Import Sample Data
This notebook imports a Python module used to share sample datasets.

##### Objectives
- Install Python library using %pip
- Import a module from workspace files

## Importing Sample Datasets

The `test_data_setup` module was created to modularize the code that prepared sample records for our datasets. We will import this module to get the sample data and description using this `test_data` class defined [here](https://github.com/databricks-academy/advanced-data-engineering-test-data-example/blob/published/src/data_setup/test_data_setup.py). This module was packaged in a library and released as a Python wheel file [here](https://github.com/databricks-academy/advanced-data-engineering-test-data-example/releases), which can be installed using the `%pip` command. Alternatively, this module can be stored as relative library using Files in Repos and imported directly using relative paths.

## Install Python Library with %pip
In the next two cells, we generate the install command and run **`%pip`** to install the package.

**NOTE:** You should always place all **`%pip`** commands at the beginning of the notebook, since they modify the execution environment and reset the notebook state. Any Python methods or variables created in a notebook are lost when you run **`%pip`**.

````python
version = "0.0.5"
data_library_url = f"https://github.com/databricks-academy/advanced-data-engineering-test-data-example/releases/download/v{version}/data_setup-{version}-py3-none-any.whl"
pip_command = f"install --quiet --disable-pip-version-check {data_library_url}"
````

````python
%pip $pip_command
````

We can now import the module for our test data from the package installed above.

````python
from data_setup import test_data_setup
````


## Import Relative Libraries

Alternatively, we can store this module in a relative library using Files in Repos, and import the module directly using relative paths. This can be helpful for ensuring that our notebook always uses the correct version.

For example, say the module code from the wheel package installed above was refactored into a Python file stored in the same repo and directory as this notebook. If this file was called `test_data_setup.py`, we can now import this without installing any packages, with the following import statement.

```
import test_data_setup
```

## Prepare Sample Data

Let's use our imported module to access our sample datasets.

The code below uses the `test_data` class to get sample data, print one of the datasets (workouts), and calls a method to view a description of all sample datasets provided by this module.

````python
data = test_data_setup.test_data()

print(data.workouts_json)
print(data.get_data_description())
````

````python
import pandas as pd

pdf = pd.DataFrame(data.workouts_json.split("\n"), columns=["data"])
df = spark.createDataFrame(pdf)

display(df)
````

We can use this to create our sample dataset.

````python
from pyspark.sql.functions import col, from_json

schema = "user_id INT, workout_id INT, timestamp FLOAT, action STRING, session_id INT"
df = df.select(from_json(col("data"), schema).alias("json")).select("json.*")

display(df)
````
