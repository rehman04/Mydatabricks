
## Isolate Pipeline Environments

Isolate environments for dev, test, and prod pipelines by configuring different target schemas and storage locations for each environment.  
For example, following settings could be used to configure the same ETL pipeline for different environments.

##### Production
- Name: `ETL Pipeline - Prod`
- Target: `prod_schema`
- Storage: `/path/to/prod-storage`

##### Development
- Name: `ETL Pipeline - Dev`
- Target: `dev_schema`
- Storage: `/path/to/dev-storage`

##### Test
- Name: `ETL Pipeline - Test`
- Target: `test_schema`
- Storage: `/path/to/test-storage`

The development pipeline targets a development database and uses a storage location for development.  
The test pipeline targets a test database and uses a storage location for testing.  
These pipelines would be maintained in a separate environment from the production pipeline.

### Segment Ingest and Transform Libraries

We can modularize our pipeline to easily reuse the same pipeline transformation code on different data sources. 

The settings below can be used to configure libraries for pipelines that use the same transformation libraries, while using a different ingestion library to replace `<ingest-library>` below.

```
{
  "libraries": [
      "notebook": {"path": <ingest-library> },
      "notebook": {"path": ".../transform_silver.py"},
      "notebook": {"path": ".../transform_gold.py"}
  ]
}
```

Say we have three separate libraries that ingest data from different sources into bronze tables for a pipeline.
- `ingest-prod/ingest.py` loads from a kafka stream for production
- `ingest-dev/ingest-subset.py` loads a subset of production data for development
- `ingest-dev/ingest-sample.py` loads sample datasets created for unit testing

Each of the ingest libraries above define the same tables used as input data in the shared transformation library `transform_silver.py`. For example, all three libraries define a `workouts_bronze` table that is used as input for the `workouts_silver` table in the transformation library.
