
## Make API calls Using `curl`

Use the following pattern with **`curl`** to issue API calls. Note the following:
* **`cat`**, in combination with a <a href="https://tldp.org/LDP/abs/html/here-docs.html" target="_blank">here document</a>, supplies the JSON payload that specifies the API parameters; in this example, we specify the **`path`** parameter.
* the **`-H`** option, in combination with **`DATABRICKS_TOKEN`** environment variable, specifies the authentication header
* the **`DATABRICKS_HOST`** environment variable supplies the base URL for the [API endpoints](https://docs.databricks.com/api).
* **`json_pp`** formats the JSON response into a readable form.

This particular invocation lists the contents of the */Users* directory.

```python
%sh cat << EOF | curl -s -X GET -H "Authorization: Bearer $DATABRICKS_TOKEN" $DATABRICKS_HOST/api/2.0/workspace/list -d @- | json_pp
{
  "path": "/Users"
}
EOF
```


The commands are a bit simpler if we don't need to pass in parameters. For example, the following lists all clusters.

```python
%sh curl -s -X GET -H "Authorization: Bearer $DATABRICKS_TOKEN" $DATABRICKS_HOST/api/2.0/clusters/list | json_pp
```


## Make API calls using Python

You can issue API calls through Python. The easiest way to do this is to install and use the [Databricks SDK](https://docs.databricks.com/en/dev-tools/sdk-python.html). This is already installed for you in this setup.

In this section, we'll see how to duplicate the two API calls from above using the Python SDK.

```python
from databricks.sdk import WorkspaceClient

# Instantiate a client object. You can specify credentials in this initialization call
# but in this setup we are relying on DATABRICKS_HOST and DATABRICKS_TARGET
# having been set up.

w = WorkspaceClient()
```


Use the API to list contents of the */Users* folder.

```python
[ i.path for i in w.workspace.list('/Users') ]
```


Use the API to list clusters.

```python
[ (i.cluster_id,i.cluster_name) for i in w.clusters.list() ]
```
